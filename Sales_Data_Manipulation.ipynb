```python
import pandas as pd
import numpy as np
```

```python
sales=pd.read_csv('sales.csv')
```

```python
sales
```

```python
sales['quantity']=sales.groupby('product')['quantity'].transform(lambda x:x.fillna(x.mean()))
```

```python
sales
```

```python
sales['price']=sales.groupby('product')['price'].transform(lambda x:x.fillna(x.median()))
```

```python
sales
```

```python
sales['date']=pd.to_datetime(sales['date'])
```

```python
sales['month']=sales['date'].dt.month
```

```python
sales
```

```python
sales['total sales']=sales['quantity']*sales['price']
```

```python
sales
```

```python
sales
```

```python
mean_quantity=np.mean(sales['quantity'])
```

```python
median_quantity=np.median(sales['quantity'])
```

```python
std_quantity=np.std(sales['quantity'])
```

```python
mean_price=np.mean(sales['price'])
```

```python
median_price=np.median(sales['price'])
```

```python
std_price=np.std(sales['price'])
```

```python
print(f'the quantity mean {mean_quantity}, the quantity median {median_quantity}, the quantity_std {std_quantity}')
```

```python
print(f'the price mean {mean_price},  median {median_price}, std {std_price}')
```

```python
grouped=sales.groupby('product').agg({'quantity': 'sum' , 'total sales':'sum'})
```
